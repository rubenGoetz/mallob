```python
import pandas as pd
import os
import re
from matplotlib import pyplot as plt
from utils import *
```

```python
path_short = '../sMUC_logs_serial_300s'
path_long = '../sMUC_log_serial_3000s'
```

```python
df_short = parse_serial_overview(path_short, 'Serial')
df_short
```

```python
df_long = parse_serial_overview(path_long, 'Serial')
df_long
```

```python
combined_data = {
    'name': [],
    'Serial_result': [],
    'Serial_memory': [],
    'Serial_runtime': []
}

for name in df_short['name']:
    row = df_long[df_long['name'] == name]
    if not len(row):
        row = df_short[df_short['name'] == name]
    
    #print(list(row['name'])[0])

    combined_data['name'].append(list(row['name'])[0])
    combined_data['Serial_result'].append(list(row['Serial_result'])[0])
    combined_data['Serial_memory'].append(list(row['Serial_memory'])[0])
    combined_data['Serial_runtime'].append(list(row['Serial_runtime'])[0])

df = pd.DataFrame(combined_data)
df
```

```python
df.to_csv('serial_kissat.csv', sep='\t')
```
